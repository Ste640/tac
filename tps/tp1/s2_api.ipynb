# APIs: requêtes HTTP

## Imports

```python
import requests
headers = {'User-Agent': 'tac-cours-ulb (suzanne.djoufack.tsamo@ulb.be)'}
url = "https://openlibrary.org/search.json"
params = {"q": "Victor Hugo", "limit": 5}

resp = requests.get(url, params=params,  headers=headers)
print(resp.status_code)
resp.raise_for_status()
data = resp.json()

for livre in data ["docs"]:
    titre = livre.get("title")
    auteurs = ",".join(livre.get("author_name", []))
    annee = livre.get("first_publish_year")
    print(f"({auteurs}, {annee})")
```
